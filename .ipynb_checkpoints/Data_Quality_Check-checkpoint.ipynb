# Data Quality check of Data/olist_order_reviews_dataset.csv

```python
order_reviews = pd.read_csv("Data/olist_order_reviews_dataset.csv")
```

```python
order_reviews.head(10)
```

```python
order_reviews.info()
```

```python
order_reviews.shape
```

```python
order_reviews.isnull().sum()
```

```python
order_reviews_cleaned = order_reviews.copy()
order_reviews_cleaned["review_comment_title"] = order_reviews_cleaned["review_comment_title"].fillna("No Review Title")
order_reviews_cleaned["review_comment_message"] = order_reviews_cleaned["review_comment_message"].fillna("No Review Message")
```

```python
order_reviews_cleaned.duplicated().sum()
```

```python
review_date_columns = ["review_creation_date","review_answer_timestamp"]
for col in review_date_columns:
    order_reviews_cleaned[col] = pd.to_datetime(
        order_reviews_cleaned[col])
```

```python
order_reviews_cleaned.head(10)
```

```python
orders[orders["order_approved_at"].isnull()]["order_status"].value_counts()
```

```python
orders[orders["order_delivered_carrier_date"].isnull()]["order_status"].value_counts()
```

```python
orders[(orders["order_delivered_carrier_date"].isnull()) & (orders["order_status"] == "delivered")]
```

```python
orders[orders["order_delivered_customer_date"].isnull()]["order_status"].value_counts()
```

```python
orders[(orders["order_delivered_customer_date"].isnull()) & (orders["order_status"] == "delivered")]
```

```python
date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"]
for col in date_columns:
    orders[col] = pd.to_datetime(
        orders[col])
```

```python
orders.head(5)
```

```python
orders.duplicated().sum()
```

### Missing Values Decision

Missing values in the order date columns were analyzed based on order status.
Most missing values are logically associated with orders that were canceled,
unavailable, processing, invoiced, or not yet delivered.

Therefore, the missing values were retained and not replaced or removed to
avoid losing valid order records.

# Data Quality check of Data/olist_products_dataset.csv

```python
products = pd.read_csv("Data/olist_products_dataset.csv")
```

```python
products.head(5)
```

```python
products.info()
```

```python
products.shape
```

```python
products.columns
```

```python
products.isnull().sum()
```

```python
products[products["product_category_name"].isnull()].head(5)
```

```python
products[products["product_category_name"].isnull()].shape
```

```python
products[products["product_category_name"].isnull()].isnull().sum()
```

```python
products["product_category_name"] = products["product_category_name"].fillna("Unknown")
```

```python
products["product_category_name"].isnull().sum()
```

```python
products.isnull().sum()
```

```python
products["product_name_lenght"] = products["product_name_lenght"].fillna(products["product_name_lenght"].median())
```

```python
products["product_name_lenght"].isnull().sum()
```

```python
products["product_description_lenght"] = products["product_description_lenght"].fillna(products["product_description_lenght"].median())
```

```python
products["product_description_lenght"].isnull().sum()
```

```python
products["product_photos_qty"] = products["product_photos_qty"].fillna(products["product_photos_qty"].median())
```

```python
products["product_photos_qty"].isnull().sum()
```

```python
products.isnull().sum()
```

```python
products.dropna(  subset=[
        "product_weight_g",
        "product_length_cm",
        "product_height_cm",
        "product_width_cm"
    ],inplace=True)
```

```python
products.isnull().sum()
```

```python
products.duplicated().sum()
```
