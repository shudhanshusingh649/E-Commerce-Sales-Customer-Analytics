# Data Quality check of Data/olist_order_reviews_dataset.csv

```python
order_reviews = pd.read_csv("Data/olist_order_reviews_dataset.csv")
```

```python
order_reviews.head(10)
```

```python
order_reviews.info()
```

```python
order_reviews.shape
```

```python
order_reviews.isnull().sum()
```

```python
order_reviews_cleaned = order_reviews.copy()
order_reviews_cleaned["review_comment_title"] = order_reviews_cleaned["review_comment_title"].fillna("No Review Title")
order_reviews_cleaned["review_comment_message"] = order_reviews_cleaned["review_comment_message"].fillna("No Review Message")
```

```python
order_reviews_cleaned.duplicated().sum()
```
